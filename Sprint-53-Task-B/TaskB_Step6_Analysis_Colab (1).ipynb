|---|---|
| 1 | Upload your final Stage 2 checkpoint `sprint53_taskB_run_20261003_2105.zip`, then `sprint53_taskB_step6_code.zip` |
| 2 | Run the Stage 2 analysis (prints every table + the pre-registered decision) |
| 3 | Download the analysis tables + a checkpoint |


```python
# 1. Upload and unpack (one file at a time)
import os, zipfile
from google.colab import files
os.chdir('/content')

def unpack(path):
    with zipfile.ZipFile(path) as z:
        tops = {p.split('/')[0] for p in z.namelist() if p.strip('/')}
        if tops != {'sprint53'}:
            raise SystemExit(f'!! {path} does not contain a single sprint53/ folder (found {tops}).')
        z.extractall('/content')
    print('   unpacked', path)

have_data = lambda: os.path.exists('/content/sprint53/results_taskB/B2_wording_pairs.jsonl')
have_code = lambda: os.path.exists('/content/sprint53/taskB/analyze_stage2.py')
code_zips = []
while not (have_data() and have_code()):
    need = []
    if not have_data(): need.append('final Stage 2 checkpoint (sprint53_taskB_run_20261003_2105.zip)')
    if not have_code(): need.append('Step B6 code (sprint53_taskB_step6_code.zip)')
    print('Please upload ONE file now:', ' / '.join(need))
    for name in files.upload():
        unpack(name)
        if 'step6_code' in name.lower(): code_zips.append(name)
        else:
            for c in code_zips: unpack(c)
os.chdir('/content/sprint53')
print('\nReady.')
```

```python
# 3. Download the tables and a checkpoint
import datetime, shutil
from google.colab import files
shutil.make_archive('/content/stage2_tables', 'zip', 'analysis_taskB/stage2')
files.download('/content/stage2_tables.zip')
stamp = datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%d_%H%M')
out = f'/content/sprint53_taskB_B6_{stamp}.zip'
with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as z:
    for root, dirs, fnames in os.walk('/content/sprint53'):
        dirs[:] = [d for d in dirs if d not in ('node_modules', '.venv', '__pycache__')]
        for f in fnames:
            if f.endswith('.zip'): continue
            full = os.path.join(root, f); z.write(full, os.path.relpath(full, '/content'))
print(f'{out} ({os.path.getsize(out)/1e6:.1f} MB)')
files.download(out)
```
